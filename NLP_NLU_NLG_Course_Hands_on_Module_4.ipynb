```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

# 1. Training Data: Giving the "Sorting Hat" examples
# We provide the text and the "Label" (1 for Positive, 0 for Negative)
reviews = [
    "I loved this movie, it was amazing!",
    "Total waste of time, I hated it.",
    "The acting was great and the story was beautiful.",
    "Very boring and slow, I do not recommend it.",
    "Simply the best film of the year!",
    "The acting was terrible, a complete disaster."
]
labels = [1, 0, 1, 0, 1, 0]

# 2. Preparation: Turning our reviews into a Bag of Words
vectorizer = CountVectorizer()
review_counts = vectorizer.fit_transform(reviews)

# 3. Training: The machine studies the patterns
# It notices that "loved" and "amazing" usually belong to Label 1
classifier = MultinomialNB()
classifier.fit(review_counts, labels)

# 4. The Test: Give it a new sentence it has NEVER seen before
test_review = ["I hated the acting, it was very boring"]
test_counts = vectorizer.transform(test_review)
prediction = classifier.predict(test_counts)

# 5. The Result
sentiment = "Positive" if prediction[0] == 1 else "Negative"
print(f"Review: '{test_review[0]}'")
print(f"The Robot says this is: {sentiment}")
```
