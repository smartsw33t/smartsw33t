## Install Ollama and a Local LLM

Ollama makes it easy to run large language models locally. It handles model downloads, setup, and provides a simple API to interact with them.

First, you need to install Ollama on your system. Since Colab runs on a remote Linux environment, we'll use a `curl` command to download and install it. This will set up the Ollama server.

```python
# Install zstd, a compression utility required by Ollama
!sudo apt-get update && sudo apt-get install -y zstd

# Now, install Ollama
!curl -fsSL https://ollama.com/install.sh | sh
```

Next, we'll start the Ollama server in the background. This is crucial for the `ollama` commands to work.

```python
!nohup /usr/local/bin/ollama serve &
```

Now that Ollama is running, you can pull a model. Let's use `llama2` as an example. This might take some time depending on the model size and your internet connection. You can replace `llama2` with other models available on the [Ollama library](https://ollama.com/library).

```python
# Give Ollama a moment to start up (optional, but can prevent issues)
import time
time.sleep(5)

!/usr/local/bin/ollama pull llama2
```

Finally, you can interact with the model using Ollama's Python client. First, install the client library, then write a simple script to generate text.

```python
!pip install ollama
```

```python
import ollama
import time

# Give Ollama server and model load time to initialize
time.sleep(20)

response = ollama.chat(model='llama2',
                        messages=[
                            {
                                'role': 'user',
                                'content': 'Why is the sky blue?',
                            },
                        ])
print(response['message']['content'])
```
