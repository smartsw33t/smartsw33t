```python
import spacy
from transformers import pipeline

# 1. Setting up the "Brain" and the "Voice"
# We load a linguistics model for analysis (NLU) and a generative model for speech (NLG).
# 'en_core_web_sm' is a small, efficient English brain.
nlp_brain = spacy.load("en_core_web_sm")

# 'gpt2' is a famous generative model that knows how to construct human sentences.
voice_box = pipeline("text-generation", model="gpt2")

def smart_concierge(guest_message):
    # --- STEP A: NLU - Understanding the VIPs (Room numbers/Items) ---
    # We parse the message to find specific numerical entities.
    doc = nlp_brain(guest_message)
    room = "unknown"

    # We loop through the entities spotted by the brain
    for ent in doc.ents:
        # 'CARDINAL' is the software label for specific numbers
        if ent.label_ == "CARDINAL":
            room = ent.text

    # --- STEP B: NLG - Crafting the personalized reply ---
    # We use a 'F-String' to inject our discovered data directly into the robot's prompt.
    # This guides the model to stay on topic.
    prompt = f"The guest in Room {room} wants a drink. The hotel staff replies: 'Certainly, we will"

    # The voice box generates a continuation of our prompt
    reply = voice_box(prompt, max_length=40, num_return_sequences=1)

    return reply[0]['generated_text']

# --- The Moment of Truth ---
# Let's test our integrated system
message = "I am in Room 405 and I would like some coffee."
final_response = smart_concierge(message)

print("--- The Concierge System in Action ---")
print(f"Guest Message: {message}")
print(f"Robot Response: {final_response}")
```
