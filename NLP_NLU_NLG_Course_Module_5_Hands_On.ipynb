```python
import spacy
from spacy import displacy

# 1. Loading the Brain: We give the software a pre-trained English model
# 'sm' stands for 'small'—it's fast and efficient for this task.
nlp_brain = spacy.load("en_core_web_sm")

# 2. The Crowd: A sentence full of people, places, and organizations
raw_text = "Jayaseelan Samuel visited the Google office in Hyderabad last Tuesday."

# 3. The Scanning: The software processes the text into a 'Doc' object
# This is where the actual identification happens.
doc = nlp_brain(raw_text)

# 4. The Reveal: Iterating through the entities found
print("--- The VIPs Found ---")
for entity in doc.ents:
    print(f"Entity: {entity.text} | Label: {entity.label_}")

# 5. Visualizing: This creates a beautiful highlighted view in your browser
displacy.render(doc, style="ent", jupyter=True)
```
