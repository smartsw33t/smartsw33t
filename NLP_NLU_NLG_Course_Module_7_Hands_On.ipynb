```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# 1. Load the tokenizer and model directly for summarization
tokenizer = AutoTokenizer.from_pretrained("sshleifer/distilbart-cnn-12-6")
model = AutoModelForSeq2SeqLM.from_pretrained("sshleifer/distilbart-cnn-12-6")

# 2. The Long Story: A paragraph with lots of detail (Unstructured Data)
long_text = """
Natural Language Processing is a fascinating field of Artificial Intelligence.
It allows computers to understand human language, which was once thought impossible.
By using complex math and huge amounts of data, we can now teach machines to read,
write, and even translate between different languages like Tamil and English.
This technology is used in everything from search engines to voice assistants
that live in our pockets.
"""

# 3. Encode the text and generate the summary
inputs = tokenizer([long_text], max_length=1024, return_tensors='pt', truncation=True)

# Generate Summary using the model's generate method
summary_ids = model.generate(
    inputs['input_ids'],
    max_length=30,
    min_length=10,
    num_beams=4,
    early_stopping=True
)

# 4. Decode and Reveal the summary
summary_text = [tokenizer.decode(g, skip_special_tokens=True, clean_up_tokenization_spaces=False) for g in summary_ids]

print("--- The Essence ---")
print(summary_text[0])
```
