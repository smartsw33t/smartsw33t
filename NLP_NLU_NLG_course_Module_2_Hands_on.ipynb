```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

# Download the specific "books" our tools need
nltk.download('punkt')
nltk.download('stopwords')

print("Tools are ready for the harvest!")
```

```python
sentence = "The gardeners are planting beautiful flowers in the gardens while walking happily."

# 1. Tokenization (Separating Seeds)
tokens = word_tokenize(sentence)
print("1. Tokens:", tokens)

# 2. Stopword Removal (Pulling Weeds)
stop_words = set(stopwords.words('english'))
clean_tokens = [w for w in tokens if w.lower() not in stop_words]
print("2. Cleaned:", clean_tokens)

# 3. Stemming (Finding Roots)
ps = PorterStemmer()
roots = [ps.stem(w) for w in clean_tokens]
print("3. Ancient Roots:", roots)
```
