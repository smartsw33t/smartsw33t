```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

# 1. Our Collection of Baskets (Sentences)
baskets = [
    "The garden is full of beautiful red roses.",
    "Roses are red and violets are blue.",
    "The gardener is watering the red roses in the garden."
]

# 2. The TF-IDF Lab: Preparing the digital scale
scale = TfidfVectorizer()

# 3. Weighing the Words: Turning text into numbers
matrix = scale.fit_transform(baskets)

# 4. Creating a readable map (Table)
feature_names = scale.get_feature_names_out()
importance_map = pd.DataFrame(matrix.toarray(), columns=feature_names)

print("--- The Importance Map (TF-IDF Weights) ---")
print(importance_map)
```
