```python
from transformers import pipeline

# 1. The Factory: We are choosing a "text-generation" task.
# This initializes a model (GPT-2) that has been pre-trained on a massive
# library of internet text.
voice_box = pipeline("text-generation", model="gpt2")

# 2. The Seed: We give the robot a "Prompt" to start its thought.
# In NLG, the prompt is the 'Initial State' that guides the generation.
prompt = "In the heart of the ancient garden, there was a"

# 3. The Generation: The robot "breathes" and finishes the sentence.
# 'max_length' sets the limit for the output tokens.
output = voice_box(prompt, max_length=30, num_return_sequences=1)

# 4. The Reveal
print("--- The Robot's Story ---")
print(output[0]['generated_text'])
```
