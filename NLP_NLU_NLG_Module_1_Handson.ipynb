```python
import string

# 1. The Raw Soil: Messy text with noise
text = "  OH MY GOODNESS!!! I am... so HAPPY to learn NLP today! #AI  "

# 2. Shaking off the dirt (Removing extra spaces)
clean_text = text.strip()

# 3. Making the ground level (Turning to lowercase)
clean_text = clean_text.lower()

# 4. Removing the stones (Sieving out punctuation)
clean_text = clean_text.translate(str.maketrans('', '', string.punctuation))

print(clean_text)
```
